```python
import torch
import torch.nn as nn
from norme_spectrale_debruiteur import power_method_q


class ScalingModel(nn.Module):
    def __init__(self, scale):
        super().__init__()
        self.scale = scale

    def forward(self, image, sigma):
        return self.scale * image


def test_power_method():
    image = torch.randn(1, 4, 16, 16)
    sigma = torch.tensor(0.1)

    for scale in [0.25, 1.0, 2.0]:
        model = ScalingModel(scale)

        estimated = power_method_q(
            model,
            image,
            sigma,
            n_iter=5,
            epsilon=1e-3,
        ).item()

        expected = abs(2 * scale - 1)

        print(
            f"scale={scale} | "
            f"attendu={expected:.4f} | "
            f"obtenu={estimated:.4f}"
        )

        assert abs(estimated - expected) < 1e-2


if __name__ == "__main__":
    test_power_method()
```
